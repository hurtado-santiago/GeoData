```python
from shapely.geometry import Point, Polygon    # to create Polygons and Points
import folium    # To create map
import geopandas as gpd    # To locate inside the map
import pandas as pd    # To process data
```

```python
df = pd.read_pickle('geo_veredas2.pkl')    # Read the file in .pkl format to
```

```python
# Center the map in Chaparral
m = folium.Map(location=[3.7728555555591194, -75.57493008952609], tiles="Stamen Toner", zoom_start=10)
```

```python
lon_point_list = []
lat_point_list = []

for name, points in df.iterrows():
    for values in points[0]:
        lon_point_list.append(values[0])
        lat_point_list.append(values[1])

    polygon_geom = Polygon(zip(lon_point_list, lat_point_list))    # Polygon(x, y)
    lon_point_list = []
    lat_point_list = []
    crs = {'init': 'epsg:4326'}
    polygon = gpd.GeoDataFrame(index=[0], crs=crs, geometry=[polygon_geom])
    folium.GeoJson(polygon, name='chaparral').add_to(m)
```
